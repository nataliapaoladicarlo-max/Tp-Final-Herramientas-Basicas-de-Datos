# Visualizacion

## Graficos

## Cantidad de Matrimonios por Minucupios

```python
print(set.columns)
```

```python
import matplotlib.pyplot as plt
```

```python
set.groupby('municipio_nombre')['cantidad'].sum().sort_values(ascending=False).head(10).plot(kind='bar')

plt.title('Top 10 Municipios por Cantidad')
plt.xlabel('Municipio')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)

plt.show()
```

## Cantidad Por Mes

```python
import seaborn as sns
```

```python
set.groupby('mes')['cantidad'].sum().sort_index().plot(kind='bar')

plt.title('Cantidad por Mes')
plt.xlabel('Mes')
plt.ylabel('Cantidad')

plt.show()
```

```python
set.groupby('mes')['cantidad'].sum().sort_index().plot(kind='line', marker='o')

plt.title('Evolución de Cantidad por Mes')
plt.xlabel('Mes')
plt.ylabel('Cantidad')

plt.show()
```

```python
top10 = set.groupby('municipio_nombre')['cantidad'].sum().sort_values(ascending=False).head(10)

plt.bar(top10.index, top10.values)

plt.title('Top 10 Municipios por Cantidad')
plt.xlabel('Municipio')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)

plt.show()
```

```python
top3 = set.groupby('municipio_nombre')['cantidad'].sum().sort_values(ascending=False).head(3)

top3.plot(
    kind='pie',
    labels=top3.index,
    autopct='%1.1f%%',
    figsize=(7,7)
)

plt.title('Top 3 municipios con mayor cantidad de matrimonios')
plt.ylabel('')

plt.show()
```

## Sintesis Final

El análisis de los registros de matrimonios durante 2026 permitió identificar diferencias significativas entre municipios y períodos del año. Los municipios con mayor cantidad de matrimonios fueron Lomas de Zamora, La Plata y La Matanza, concentrando una proporción importante de los registros. Asimismo, se observó una disminución de matrimonios durante los meses de invierno, especialmente en julio y agosto, seguida de una recuperación hacia el final del año. Las visualizaciones facilitaron la identificación de patrones geográficos y temporales, permitiendo comprender mejor la distribución de los matrimonios en la provincia analizada.
